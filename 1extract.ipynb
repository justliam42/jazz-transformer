```python

def unobfusc50(s: str) -> str:
    chars = list(s)

    # Reverse positions 1–5 with 46–50
    for i in range(5):
        chars[i], chars[50 - 1 - i] = chars[50 - 1 - i], chars[i]

    # Reverse positions 11–24 with 27–40
    for i in range(10, 24):
        chars[i], chars[50 - 1 - i] = chars[50 - 1 - i], chars[i]

    return "".join(chars)


def unobfusc(s: str) -> str:
    result = ""

    while len(s) > 50:
        block = s[:50]
        s = s[50:]

        if len(s) < 2:
            result += block
        else:
            result += unobfusc50(block)

    result += s
    return result
```
